## Phần 3: Tạo cột mới và biến đổi dữ liệu


**Câu 10.** Từ cột `ORDERDATE`, tạo thêm 3 cột mới: `ORDER_YEAR`, `ORDER_MONTH`, `ORDER_QUARTER` chứa năm, tháng, quý của ngày đặt hàng.

```python
df['ORDER_YEAR'] = df.ORDERDATE.dt.year # chen them cot nam
df.ORDERDATE.dt.year # liêt kê ra các năm
df['ORDER_MONTH'] = df.ORDERDATE.dt.month
df['ORDER_QUARTER'] = df.ORDERDATE.dt.quarter
df
```

**Câu 11.** Tạo cột mới `PROFIT_MARGIN` = `Profit` / `SALES` (tỷ suất lợi nhuận). Đơn hàng nào có tỷ suất lợi nhuận cao nhất? Thấp nhất?

```python
df['PROFIT_MARGIN'] = df["Profit"] / df['SALES']
df
```

**Câu 12.** Tạo cột `NET_SALES` = `SALES` * (1 - `Discount`) thể hiện doanh số sau khi trừ giảm giá. So sánh tổng `SALES` và tổng `NET_SALES` của toàn bộ dữ liệu.
